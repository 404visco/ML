```python
#Pisah
X= data.copy()
y= X.pop("SalePrice")
```

```python
#
```

```python
#Mutual Information

#Encode data tipe non angka 
for col in X.select_dtypes("object"):
    X[col], _ = X[col].factorize()

cat_col= X.dtypes == int
```

```python
def make_mi_scores(X, y, cat_col):
    mi_scores = mutual_info_regression(X,y,discrete_features= cat_col) #Buat Skornya
    mi_scores= pd.Series(mi_scores, name= "MI Scores", index= X.columns) #Buat Datanya
    mi_scores= mi_scores.sort_values(ascending=True) #Urutin
    return mi_scores

mi_scores= make_mi_scores(X, y, cat_col)

mi_scores[::3]
```
